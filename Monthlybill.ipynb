```python
from datetime import datetime, timedelta
from collections import defaultdict
from dateutil.relativedelta import relativedelta

def generate_monthly_bill(item_list, target_month):
    start_of_month = datetime.strptime(target_month, "%Y-%m")
    end_of_month = (start_of_month + relativedelta(months=1)) - timedelta(days=1)

    grouped_items = defaultdict(list)

    for item in item_list:
        key = (item["item_code"], item["rate"], item["billing_period"])
        grouped_items[key].append(item)

    line_items = []
    total_amount = 0

    for key, group in grouped_items.items():
        item_code, rate, billing_period = key
        for item in group:
            # Parse string dates to datetime objects
            start_date = datetime.strptime(item["start_date"], "%Y-%m-%d")
            end_date = datetime.strptime(item["end_date"], "%Y-%m-%d")

            active_start = max(start_date, start_of_month)
            active_end = min(end_date, end_of_month)

            if active_end >= active_start:
                num_days = (active_end - active_start).days + 1
                amount = num_days * rate
                total_amount += amount
                line_items.append({
                    "item_code": item_code,
                    "rate": rate,
                    "billing_period": billing_period,
                    "num_days": num_days,
                    "amount": amount
                })

    return {
        "line_items": line_items,
        "total_revenue": total_amount
    }

# Correct input with dates as strings
item_list = [
    {
        "item_code": "A123",
        "rate": 10,
        "billing_period": "monthly",
        "start_date": "2025-03-15",
        "end_date": "2025-04-25"
    },
    {
        "item_code": "B456",
        "rate": 20,
        "billing_period": "monthly",
        "start_date": "2025-04-01",
        "end_date": "2025-04-30"
    },
    {
        "item_code": "C789",
        "rate": 15,
        "billing_period": "monthly",
        "start_date": "2025-03-05",
        "end_date": "2025-04-05"
    }
]

target_month = "2025-04"

monthly_bill = generate_monthly_bill(item_list, target_month)
print(monthly_bill)
```
